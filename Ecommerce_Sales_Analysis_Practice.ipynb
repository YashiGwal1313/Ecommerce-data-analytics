## 🧠 Business Analysis

The dataset was analyzed to answer important business questions
related to sales, profit, products, categories, cities, and
payment modes.

```python
 # Q1. What is the total sales?
df['Sales'].sum()
```

```python
# Q2. What is the total profit?
df['Profit'].sum()
```

```python
# Q3. What is the total quantity sold?
df['Quantity'].sum()
```

```python
# Q4. Which category generated the highest sales?
df.groupby('Category')['Sales'].sum()
```

```python
# Q5. Which city generated the highest sales?
df.groupby('City')['Sales'].sum()
```

```python
# Q6. Which product generated the highest sales?
df.groupby('Product')['Sales'].sum()
```

```python
# Q7. What are the top 5 products by sales?
df.groupby("Product")["Sales"].sum().sort_values(ascending=False).head(5)
```

```python
# Q8. What are the monthly sales?
df.groupby("Month")["Sales"].sum()
```

```python
# Q9. Which product has the highest profit margin?
p = df.groupby('Product')[['Sales','Profit']].sum()
p['Margin'] = (p['Profit'] / p['Sales']) * 100
p['Margin'].sort_values(ascending=False)
```

```python
# Q10. Which city generated the highest profit?
df.groupby('City')['Profit'].sum().sort_values(ascending=False)
```

```python
# Q11. Which payment mode generated the highest profit?
df.groupby('Payment_Mode')['Profit'].sum().sort_values(ascending=False)
```

```python
# Q12. Which category generated the highest profit?
df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
```

```python
# Q13. Which product generated the highest quantity sold?
df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)
```

```python
# Q14. Which city has the highest average sales per order?
df.groupby('City')['Sales'].mean().sort_values(ascending=False)
```

```python
# Q15.Which category has the highest average profit per order?
df.groupby('Category')['Profit'].mean().sort_values(ascending=False)
```

## 📈 Data Visualization

Visualizations were created using Matplotlib to identify patterns,
trends, and comparisons in the sales data.

### 📊 Sales by Category

```python
import matplotlib.pyplot as plt
category_sales = df.groupby('Category')['Sales'].sum()
category_sales.plot(kind='bar')

plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.show()
```

### 📊 Sales by City

```python
city_sales = df.groupby('City')['Sales'].sum()
city_sales.plot(kind='bar')

plt.title('Sales by City')
plt.xlabel('City')
plt.ylabel('Total Sales')
plt.show()
```

### 📊 Top 5 Products by Sales

```python
top5_products = df.groupby('Product')['Sales'].sum().sort_values(ascending=False).head(5)
top5_products.plot(kind='bar')

plt.title('Top 5 Products by Sales')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

### 📈 Monthly Sales Trend

```python
monthly_sales = df.groupby('Month')['Sales'].sum()
monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.show()
```

### 📊 Sales by Payment Mode

```python
payment_sales = df.groupby('Payment_Mode')['Sales'].sum()
payment_sales.plot(kind='bar')

plt.title('Sales by Payment Mode')
plt.xlabel('Payment Mode')
plt.ylabel('Total Sales')
plt.show()
```
